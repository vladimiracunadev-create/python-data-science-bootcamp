## 1️⃣ Tipos primitivos y mutabilidad

| Tipo | Mutable | Hashable | Uso típico |
|---|---|---|---|
| `int`, `float`, `complex` | ❌ | ✅ | aritmética |
| `bool` | ❌ | ✅ | flags |
| `str` | ❌ | ✅ | texto |
| `bytes` | ❌ | ✅ | binario |
| `bytearray` | ✅ | ❌ | binario mutable |
| `list` | ✅ | ❌ | secuencia ordenada |
| `tuple` | ❌ | ✅ (si elementos hashables) | record, return múltiple |
| `set`, `frozenset` | ✅/❌ | ❌/✅ | unicidad, operaciones de conjunto |
| `dict` | ✅ | ❌ | mapping key→value, O(1) lookup |

**Regla**: solo los **hashables** pueden ser claves de dict o elementos de set.

````python
# Demostrar mutabilidad: shared reference vs copy
a = [1, 2, 3]
b = a            # MISMO objeto
b.append(4)
print('a:', a)   # [1, 2, 3, 4]  ← cambió

c = a.copy()     # NUEVO objeto
c.append(99)
print('a:', a)   # sin cambio
print('c:', c)
````

## 2️⃣ Las 4 estructuras — cuándo cada una

- **list** — orden importa, duplicados OK, append/pop por el final.
- **tuple** — inmutable, hashable, ideal para *records* y returns múltiples.
- **set** — unicidad y operaciones (unión, intersección, diferencia) en O(1).
- **dict** — mapping key→value, lookup O(1), preserva orden de inserción (Python 3.7+).

````python
# Comparativa práctica
frutas = ['manzana', 'pera', 'manzana', 'kiwi']

# Unicidad rápida con set
print('únicas:', set(frutas))

# Conteo con dict
conteo = {}
for f in frutas:
    conteo[f] = conteo.get(f, 0) + 1
print('conteo:', conteo)

# Record con tuple (inmutable)
punto = (3.5, 7.1)
x, y = punto   # unpacking
print(f'x={x}, y={y}')
````

## 3️⃣ Unpacking — el superpoder olvidado

```python
a, b, c = [1, 2, 3]              # básico
a, *rest = [1, 2, 3, 4]          # rest = [2, 3, 4]
*init, last = [1, 2, 3, 4]       # init = [1, 2, 3], last = 4
for i, val in enumerate(['a','b','c']):
    print(i, val)
for k, v in {'x': 1, 'y': 2}.items():
    print(k, v)
```

````python
# Unpacking en returns y zip
def min_max(xs):
    return min(xs), max(xs)

lo, hi = min_max([5, 2, 9, 1, 7])
print(f'rango: [{lo}, {hi}]')

nombres = ['Ana', 'Bob', 'Cris']
edades  = [30, 25, 28]
for n, e in zip(nombres, edades):
    print(f'{n} tiene {e} años')
````

## 4️⃣ Truthiness — qué es "falso"

Falsy: `False`, `None`, `0`, `0.0`, `''`, `[]`, `{}`, `set()`, `range(0)`.
Todo lo demás es truthy.

**Idioma pythonico**: `if items:` en vez de `if len(items) > 0:`.

````python
for v in [0, '', [], {}, None, False, 'x', [0], {'a':1}]:
    print(f'{repr(v):>15} → {"truthy" if v else "falsy"}')
````

## 5️⃣ ⚠️ El bug del default mutable

```python
def add(item, target=[]):   # ← TRAMPA
    target.append(item)
    return target

add('a')   # ['a']
add('b')   # ['a', 'b']  ← ¡compartió la lista!
```

**Por qué**: el default se evalúa **una sola vez**, al definir la función. Esa lista vive entre llamadas.

**Fix**: usa `None` como sentinel.

````python
# El bug en vivo
def add_bug(item, target=[]):
    target.append(item)
    return target

print(add_bug('a'))
print(add_bug('b'))   # ¡compartió!
print(add_bug('c'))

# Fix
def add_ok(item, target=None):
    if target is None:
        target = []
    target.append(item)
    return target

print()
print(add_ok('a'))
print(add_ok('b'))   # cada llamada arranca limpio
print(add_ok('c'))
````
