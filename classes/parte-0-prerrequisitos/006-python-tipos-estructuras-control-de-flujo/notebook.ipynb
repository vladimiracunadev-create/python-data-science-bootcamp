## 1️⃣ Tipos primitivos y mutabilidad

| Tipo | Mutable | Hashable | Uso típico |
|---|---|---|---|
| `int`, `float`, `complex` | ❌ | ✅ | aritmética |
| `bool` | ❌ | ✅ | flags |
| `str` | ❌ | ✅ | texto |
| `bytes` | ❌ | ✅ | binario |
| `bytearray` | ✅ | ❌ | binario mutable |
| `list` | ✅ | ❌ | secuencia ordenada |
| `tuple` | ❌ | ✅ (si elementos hashables) | record, return múltiple |
| `set`, `frozenset` | ✅/❌ | ❌/✅ | unicidad, operaciones de conjunto |
| `dict` | ✅ | ❌ | mapping key→value, O(1) lookup |

**Regla**: solo los **hashables** pueden ser claves de dict o elementos de set.

````python
# Demostrar mutabilidad: shared reference vs copy
a = [1, 2, 3]
b = a            # MISMO objeto
b.append(4)
print('a:', a)   # [1, 2, 3, 4]  ← cambió

c = a.copy()     # NUEVO objeto
c.append(99)
print('a:', a)   # sin cambio
print('c:', c)
````

## 2️⃣ Las 4 estructuras — cuándo cada una

- **list** — orden importa, duplicados OK, append/pop por el final.
- **tuple** — inmutable, hashable, ideal para *records* y returns múltiples.
- **set** — unicidad y operaciones (unión, intersección, diferencia) en O(1).
- **dict** — mapping key→value, lookup O(1), preserva orden de inserción (Python 3.7+).

````python
# Comparativa práctica
frutas = ['manzana', 'pera', 'manzana', 'kiwi']

# Unicidad rápida con set
print('únicas:', set(frutas))

# Conteo con dict
conteo = {}
for f in frutas:
    conteo[f] = conteo.get(f, 0) + 1
print('conteo:', conteo)

# Record con tuple (inmutable)
punto = (3.5, 7.1)
x, y = punto   # unpacking
print(f'x={x}, y={y}')
````

## 3️⃣ Unpacking — el superpoder olvidado

```python
a, b, c = [1, 2, 3]              # básico
a, *rest = [1, 2, 3, 4]          # rest = [2, 3, 4]
*init, last = [1, 2, 3, 4]       # init = [1, 2, 3], last = 4
for i, val in enumerate(['a','b','c']):
    print(i, val)
for k, v in {'x': 1, 'y': 2}.items():
    print(k, v)
```

````python
# Unpacking en returns y zip
def min_max(xs):
    return min(xs), max(xs)

lo, hi = min_max([5, 2, 9, 1, 7])
print(f'rango: [{lo}, {hi}]')

nombres = ['Ana', 'Bob', 'Cris']
edades  = [30, 25, 28]
for n, e in zip(nombres, edades):
    print(f'{n} tiene {e} años')
````

## 4️⃣ Truthiness — qué es "falso"

Falsy: `False`, `None`, `0`, `0.0`, `''`, `[]`, `{}`, `set()`, `range(0)`.
Todo lo demás es truthy.

**Idioma pythonico**: `if items:` en vez de `if len(items) > 0:`.

````python
for v in [0, '', [], {}, None, False, 'x', [0], {'a':1}]:
    print(f'{repr(v):>15} → {"truthy" if v else "falsy"}')
````

## 5️⃣ ⚠️ El bug del default mutable

```python
def add(item, target=[]):   # ← TRAMPA
    target.append(item)
    return target

add('a')   # ['a']
add('b')   # ['a', 'b']  ← ¡compartió la lista!
```

**Por qué**: el default se evalúa **una sola vez**, al definir la función. Esa lista vive entre llamadas.

**Fix**: usa `None` como sentinel.

````python
# El bug en vivo
def add_bug(item, target=[]):
    target.append(item)
    return target

print(add_bug('a'))
print(add_bug('b'))   # ¡compartió!
print(add_bug('c'))

# Fix
def add_ok(item, target=None):
    if target is None:
        target = []
    target.append(item)
    return target

print()
print(add_ok('a'))
print(add_ok('b'))   # cada llamada arranca limpio
print(add_ok('c'))
````

## 📖 Definiciones y características

**Mutable**

Objeto cuyo estado puede modificarse después de crearlo (`list`, `dict`, `set`, `bytearray`). Consecuencia: si dos variables apuntan al mismo objeto mutable, cambiar una cambia la otra (alias).

**Inmutable**

Objeto que NO se puede modificar; cualquier 'modificación' produce un objeto nuevo (`int`, `float`, `str`, `tuple`, `frozenset`). Característica clave: son **hashables** y pueden ser claves de dict o miembros de set.

**Hashable**

Objeto con `__hash__` definido y constante durante su vida. Los inmutables built-in son hashables; las listas y dicts NO lo son. Es lo que permite el O(1) lookup de set/dict.

**Unpacking**

Sintaxis para asignar múltiples variables desde un iterable (`a, b = (1, 2)` o `a, *resto = [1, 2, 3, 4]`). El `*` captura el resto en una lista. Funciona en for, return, llamadas a funciones (`f(*lista)`, `g(**dict)`).

**Truthy / Falsy**

Cómo evalúa Python un objeto en contexto booleano (`if x:`). Falsy: `False`, `None`, `0`, `0.0`, `''`, `[]`, `{}`, `set()`. Truthy: todo lo demás (incluso `' '` con espacio, `[0]`, `{None: None}`).

## ⚠️ Errores comunes

| Síntoma / mensaje | Causa y cómo arreglar |
|---|---|
| `TypeError: unhashable type: 'list'` | Intentaste usar una lista como clave de dict o miembro de set. **Fix**: usa una tupla en su lugar (es inmutable y hashable). `{[1,2]: 'x'}` ❌ → `{(1,2): 'x'}` ✅. |
| Modifiqué una lista y otra variable también cambió | Las dos apuntan al mismo objeto (alias). **Fix**: si querías una copia, usa `list_b = list_a.copy()` o `list_b = list_a[:]` o `list_b = list(list_a)`. |
| Función con default `=[]` comparte la lista entre llamadas | El default se evalúa **una vez** al definir la función. **Fix**: `def f(x, lst=None): lst = lst or []` (o `lst = [] if lst is None else lst`). |
| `KeyError` al acceder a dict que no tiene la key | `d['x']` lanza KeyError si no existe. **Fix**: usa `d.get('x', default)` o `from collections import defaultdict; d = defaultdict(int)`. |
| `if items != None:` en vez de `if items is not None:` | Para singletons (`None`, `True`, `False`) usa siempre `is` / `is not` — más rápido y semánticamente correcto. `==` puede engañar si el objeto override `__eq__`. |

## ❓ Preguntas frecuentes

**❓ ¿Cuándo tupla y cuándo lista?**

**Tupla** para records heterogéneos de tamaño fijo (`punto = (3.5, 7.1)`, `(nombre, edad)`). **Lista** para colecciones homogéneas que crecen (`temperaturas = [22.1, 22.5, ...]`). Tupla además sirve como clave de dict; lista no.

**❓ ¿Set o dict para chequear pertenencia?**

Ambos son O(1). Set si solo necesitas saber si está. Dict si además necesitas asociar valor. Una `list` para esto es O(n) — usa set/dict si haces `if x in coleccion` con N grande.

**❓ ¿`copy()` me da una copia completa? ¿Y con listas anidadas?**

`.copy()` es **shallow**: copia el contenedor pero comparte las referencias internas. Para anidamiento, `import copy; copy.deepcopy(x)` — más lento pero independiente.

**❓ ¿Por qué `0.1 + 0.2 != 0.3`?**

Floats son IEEE 754 binarios; `0.1` y `0.2` no tienen representación exacta. **Fix**: para igualdad usa `math.isclose(a, b)`. Para precisión decimal financiera, `from decimal import Decimal`.

**❓ ¿Qué pasa con dict si insertas la misma clave dos veces?**

El segundo valor sobrescribe al primero. El orden de inserción se preserva (Python 3.7+), así que `list(d)` da las keys en el orden en que se insertaron por primera vez.

**Ejercicio 1.** Cuenta palabras: dado un texto, devolver un `dict[str, int]` con frecuencias, sin usar `Counter`.

````python
# Solución Ej.1 — frecuencia de palabras a mano (sin collections.Counter)
def contar_palabras(texto: str) -> dict:
    """O(n) sobre el número de palabras; dict.get evita el KeyError."""
    freq = {}
    for palabra in texto.lower().split():
        freq[palabra] = freq.get(palabra, 0) + 1
    return freq

texto = "el gato y el perro y el gato"
resultado = contar_palabras(texto)
print(resultado)
assert resultado == {"el": 3, "gato": 2, "y": 2, "perro": 1}
assert contar_palabras("") == {}          # edge case: texto vacío
print("OK: frecuencias correctas.")
````

**Ejercicio 2.** Unique con orden: recibir `list[int]` y devolver los únicos manteniendo el orden de primera aparición.

````python
# Solución Ej.2 — únicos preservando orden. Un set da unicidad O(1) para el "visto",
# pero el orden lo garantiza la lista de salida.
def unique_ordenado(xs: list) -> list:
    vistos = set()
    salida = []
    for x in xs:
        if x not in vistos:
            vistos.add(x)
            salida.append(x)
    return salida

datos = [3, 1, 2, 3, 1, 4, 2, 5]
print(unique_ordenado(datos))
assert unique_ordenado(datos) == [3, 1, 2, 4, 5]
assert unique_ordenado([]) == []          # edge case
# Truco 3.7+: dict.fromkeys preserva orden de inserción y da lo mismo
assert list(dict.fromkeys(datos)) == unique_ordenado(datos)
print("OK: únicos con orden de primera aparición.")
````

**Ejercicio 3.** Reproducir el bug del default mutable: `def add(item, target=[])` llamada 3 veces. Observar, explicar y arreglar.

````python
# Solución Ej.3 — el default mutable se evalúa UNA sola vez al definir la función.
def add_buggy(item, target=[]):
    target.append(item)
    return target

r1 = add_buggy("a")
r2 = add_buggy("b")
r3 = add_buggy("c")
print("buggy acumula:", r3)
# La MISMA lista se comparte entre llamadas -> se va acumulando:
assert r3 == ["a", "b", "c"]
assert r1 is r2 is r3     # las tres devuelven el MISMO objeto

# Fix idiomático: default None y crear la lista dentro
def add_ok(item, target=None):
    target = [] if target is None else target
    target.append(item)
    return target

assert add_ok("a") == ["a"]
assert add_ok("b") == ["b"]   # cada llamada arranca fresca
print("OK: con default=None cada llamada obtiene su propia lista.")
````

**Ejercicio 4.** Top-K palabras: mismo texto del ejercicio 1, devolver las 5 más frecuentes ordenadas por frecuencia descendente.

````python
# Solución Ej.4 — reutiliza contar_palabras del Ej.1 (lo redefinimos para que la celda sea autónoma).
def contar_palabras(texto: str) -> dict:
    freq = {}
    for palabra in texto.lower().split():
        freq[palabra] = freq.get(palabra, 0) + 1
    return freq

def top_k(texto: str, k: int = 5) -> list:
    freq = contar_palabras(texto)
    # ordenar por frecuencia desc; desempate estable por palabra asc
    return sorted(freq.items(), key=lambda kv: (-kv[1], kv[0]))[:k]

texto = "el gato y el perro y el gato con el raton"
resultado = top_k(texto, 3)
print(resultado)
assert resultado[0] == ("el", 4)          # 'el' es la más frecuente
assert len(top_k(texto, 100)) == len(contar_palabras(texto))  # k mayor que el vocabulario
print("OK: top-K por frecuencia descendente.")
````

**Ejercicio 5.** Grupos por inicial: dado `list[str]`, devolver `dict[str, list[str]]` agrupando por primera letra (case-insensitive).

````python
# Solución Ej.5 — agrupar por inicial. defaultdict evita el "if key not in d" repetitivo.
from collections import defaultdict

def agrupar_por_inicial(palabras: list) -> dict:
    grupos = defaultdict(list)
    for w in palabras:
        if not w:
            continue                  # ignora strings vacíos (edge case)
        inicial = w[0].lower()        # case-insensitive
        grupos[inicial].append(w)
    return dict(grupos)

datos = ["Ana", "avion", "Beto", "casa", "Carlos", "banana"]
resultado = agrupar_por_inicial(datos)
print(resultado)
assert resultado["a"] == ["Ana", "avion"]
assert resultado["b"] == ["Beto", "banana"]
assert resultado["c"] == ["casa", "Carlos"]
assert agrupar_por_inicial([]) == {}      # edge case: lista vacía
print("OK: agrupado por inicial, case-insensitive.")
````
