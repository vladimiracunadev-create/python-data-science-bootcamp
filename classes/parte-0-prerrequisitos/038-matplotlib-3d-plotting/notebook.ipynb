## ⚙️ Setup

````python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(42)
````

## 1️⃣ Habilitar 3D

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
```

## 2️⃣ Scatter 3D

````python
N = 200
xs = rng.normal(0, 1, N)
ys = rng.normal(0, 1, N)
zs = rng.normal(0, 1, N)
colors = xs*ys*zs   # 4ª dimensión codificada en color

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection='3d')
sc = ax.scatter(xs, ys, zs, c=colors, cmap='coolwarm', s=30, alpha=0.7)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
fig.colorbar(sc, label='x·y·z')
plt.title('Scatter 3D')
plt.show()
````

## 3️⃣ Superficie z = f(x, y)

````python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.9)
fig.colorbar(surf, shrink=0.5)
ax.set_title('z = sin(√(x²+y²))')
plt.show()
````

## 4️⃣ Wireframe y contour

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), subplot_kw={'projection': '3d'})
ax1.plot_wireframe(X, Y, Z, color='steelblue', alpha=0.6)
ax1.set_title('wireframe')
ax2.contour3D(X, Y, Z, levels=20, cmap='viridis')
ax2.set_title('contour3D')
plt.show()
````

## 5️⃣ Rotar con `view_init(elev, azim)`

````python
angles = [(20, 30), (20, 90), (60, 30), (10, 60)]
fig, axes = plt.subplots(2, 2, figsize=(10, 8), subplot_kw={'projection': '3d'})
for ax, (e, a) in zip(axes.flat, angles):
    ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
    ax.view_init(elev=e, azim=a)
    ax.set_title(f'elev={e}, azim={a}')
plt.tight_layout()
plt.show()
````

## 🚫 Cuándo NO usar 3D

La mayoría de las veces, un 2D bien hecho comunica mejor:

- **Scatter 3D con muchos puntos** → confuso por oclusión. Mejor: 2D + color para la 3ª dim.
- **Bar 3D** → casi siempre engaña en proporciones. Mejor: heatmap o bar agrupado 2D.
- **Pie 3D** → mentira visual. Usa bar.

**Sí usa 3D para**:
- Superficies analíticas obvias (z = f(x, y)) cuando la forma 3D **es** el mensaje.
- Datos físicos con 3 dimensiones reales (geología, partículas).
- Cuando es interactivo (rotable con `%matplotlib widget`).
