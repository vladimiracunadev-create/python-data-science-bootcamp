## ⚙️ Setup

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter, FuncFormatter
rng = np.random.default_rng(42)
```

## 1️⃣ Leyenda fuera del axes

Cuando el gráfico está saturado, sácala con `bbox_to_anchor`:

```python
x = np.linspace(0, 10, 100)

fig, ax = plt.subplots(figsize=(9, 4))
for i in range(5):
    ax.plot(x, np.sin(x + i*0.5) + i*0.3, label=f'serie {i+1}')

ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0)
ax.set_title('Leyenda fuera del axes (a la derecha)')
plt.tight_layout()
plt.show()
```

## 2️⃣ Colorbar

Cuando codificas info en color (continuo o discreto), añade colorbar:

```python
x = rng.normal(0, 1, 500)
y = rng.normal(0, 1, 500)
density = np.exp(-(x**2 + y**2)/2)

fig, ax = plt.subplots(figsize=(7, 5))
sc = ax.scatter(x, y, c=density, cmap='viridis', s=20)
cbar = fig.colorbar(sc, ax=ax)
cbar.set_label('densidad')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.set_title('Scatter coloreado por densidad')
plt.tight_layout()
plt.show()
```

## 3️⃣ Tick formatters

Cuando el eje representa porcentaje, dinero, fechas raras, etc.:

```python
categorias = ['A', 'B', 'C', 'D', 'E']
porcentajes = [0.45, 0.22, 0.15, 0.10, 0.08]   # como fracción

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(categorias, porcentajes, color='steelblue')
ax.yaxis.set_major_formatter(PercentFormatter(xmax=1.0))
ax.set_title('Distribución por categoría')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Formatter custom: USD con miles
def usd(x, pos):
    return f'${x:,.0f}'

ventas = rng.uniform(50_000, 200_000, 6)
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(range(6), ventas, color='seagreen')
ax.yaxis.set_major_formatter(FuncFormatter(usd))
ax.set_title('Ventas 2024 (USD)')
plt.tight_layout()
plt.show()
```

## 4️⃣ Anotar puntos con flecha

```python
x = np.arange(20)
y = rng.normal(50, 10, 20).cumsum()

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(x, y, marker='o', color='steelblue')

# Anotar el máximo
max_i = y.argmax()
ax.annotate(
    f'pico: {y[max_i]:.1f}',
    xy=(x[max_i], y[max_i]),                  # punto a apuntar
    xytext=(x[max_i] + 2, y[max_i] + 5),       # posición del texto
    arrowprops=dict(arrowstyle='->', color='crimson', lw=1.5),
    fontsize=11, color='crimson',
)
ax.set_title('Anotando el máximo con flecha')
plt.tight_layout()
plt.show()
```

## 5️⃣ Líneas y bandas de referencia

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(x, y, color='steelblue')

ax.axhline(y.mean(), color='gray', linestyle='--', label=f'media={y.mean():.0f}')
ax.axhspan(y.mean()-y.std(), y.mean()+y.std(), color='gray', alpha=0.15, label='±1 std')
ax.axvline(10, color='red', linestyle=':', linewidth=1, label='evento x=10')

ax.legend()
ax.set_title('Líneas y bandas de referencia')
plt.tight_layout()
plt.show()
```

## 6️⃣ Log scale

Cuando los datos cubren varios órdenes de magnitud:

```python
valores = np.array([1, 10, 100, 1_000, 10_000, 100_000])
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4))

a1.plot(valores, marker='o', color='steelblue'); a1.set_title('lineal'); a1.grid(alpha=0.3)
a2.plot(valores, marker='o', color='steelblue'); a2.set_yscale('log'); a2.set_title('log'); a2.grid(alpha=0.3, which='both')
plt.tight_layout()
plt.show()
```
